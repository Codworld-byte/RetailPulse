```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load Day 3 customer segmentation results
file_path = Path("../data/processed/customer_segments.csv")

rfm = pd.read_csv(file_path)

print("RetailPulse - Day 4")
print("Customer Insights Analysis")
print("Dataset Shape:", rfm.shape)
print("\nCustomer Segments:")
print(rfm["Segment"].value_counts())
```

```python

# Analyze customer segments
segment_summary = rfm.groupby("Segment").agg(
    Total_Customers=("Customer ID", "count"),
    Avg_Recency=("Recency", "mean"),
    Avg_Frequency=("Frequency", "mean"),
    Total_Revenue=("Monetary", "sum"),
    Avg_Spending=("Monetary", "mean")
).round(2)

segment_summary = segment_summary.sort_values(
    "Total_Revenue", ascending=False
)

display(segment_summary)
```

```python

plt.figure(figsize=(10, 6))

sns.barplot(
    data=segment_summary.reset_index(),
    x="Segment",
    y="Total_Revenue"
)

plt.title("Revenue Contribution by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Revenue")
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()
```
